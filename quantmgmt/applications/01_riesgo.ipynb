```python
from quantmgmt.data import YahooFinanceData
from quantmgmt.utils import generate_risk_summary

yf = YahooFinanceData()
```

```python
tickers = ["AAPL", "MSFT", "AMZN", "GOOGL", "NVDA"]

data = yf.download(
    tickers,
    start="2016-01-01",
    end="2025-12-31"
)
prices = data.Close
```
