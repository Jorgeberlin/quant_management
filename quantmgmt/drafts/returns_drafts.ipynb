```python
import pandas as pd
import numpy as np

prices = pd.Series(
    [100, 105, 102, 108, 110],
    index=pd.date_range("2026-01-01", periods=5)
)

print(prices)
```

```python
def cumulative_returns(prices:pd.Series | pd.DataFrame, method: str = "simple") -> pd.Series | pd.DataFrame:
    # Igual que la anterior, solamente quitamos el Nan que genera al incio.
    if method == "simple":
        returns = prices.pct_change()
        return (1 + returns).prod() - 1
    raise ValueError("method must be either 'simple' or 'log'")
```

```python
a = cumulative_returns(prices, "log")
print(a)
```
